```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
spark
```

```python
df = spark.read.csv("../files/challenge.csv", header=True)
df.show()
```

```python
import pyspark.sql.functions as sqlfunc
df2 = df.withColumn('home', sqlfunc.when(df.Country == 'Mexico', 'yes').otherwise('no'))
df2.show()
```

```python
df3 = df2.groupBy('home').agg(sqlfunc.sum(df2.Bytes_used).alias('total_used'))
df3.show()
```

```python
df4 = df2.groupBy('Country').agg(sqlfunc.countDistinct(df2.ip_address).alias('no_of_ips'))
df4.show()
```
