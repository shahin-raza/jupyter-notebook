```python
df_filter = df.filter(df.first_name.isin("Aldin", "Valma"))
df_filter.show()
```

```python
df_filter = df.filter(df.first_name.isin("Aldin", "Valma") & (df.City.like('%ondon')))
df_filter.show()
```

```python
df_filter = df.filter((df.id > 10) & (df.id < 100))
df_filter.show()
```

```python
# Sql Query
df3.createOrReplaceTempView("original")
query1 = spark.sql("SELECT * FROM original")
query1.show()
```

```python
query2 = spark.sql("SELECT CONCAT(first_name, ' ', last_name) as full_name FROM original WHERE gender='Female'")
query2.show()
```

```python
import pyspark.sql.functions as sqlfunc
df3 = df3.withColumn('are_they_female', sqlfunc.when(df3.gender == 'Female', 'Yes').otherwise('No'))
df3.show()
```

```python
# Groupby and aggregation
df3 = df3.withColumn('clean_salary', df3.Salary.substr(2, 100).cast('float'))
df3.show()
```

```python
df4 = df3.groupBy('gender').agg(sqlfunc.sum('clean_salary'))
df4.show()
```

```python
# Writing dataframe to files.
df3.write.csv('../files/df3.csv')
```
